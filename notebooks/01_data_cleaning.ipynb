# 01 — Data Cleaning

This notebook checks and cleans the raw survey CSV. Comments use simple English.

```python
from pathlib import Path
import pandas as pd

# Find the project folder and the raw data file.
project_folder = Path.cwd().parent
raw_file = project_folder / 'data' / 'raw' / 'survey_responses_raw.csv'

# Read the raw survey responses.
data = pd.read_csv(raw_file)
print('Rows and columns:', data.shape)
data.head()
```

```python
# Remove extra spaces from text answers.
for column in data.select_dtypes(include='object').columns:
    data[column] = data[column].str.strip()

# Check missing values in each column.
print(data.isna().sum())
```

```python
# Keep only people who agreed to participate.
consent_column = '1. Do you agree to participate in this survey?'
data = data[data[consent_column].str.lower() == 'yes'].copy()

# Save a cleaned copy without changing the original raw file.
clean_file = project_folder / 'data' / 'processed' / 'responses_clean.csv'
clean_file.parent.mkdir(parents=True, exist_ok=True)
data.to_csv(clean_file, index=False)
print('Cleaned file saved:', clean_file)
print('Cleaned rows:', len(data))
```
