# 03 — Suggestion Analysis

Read the open-ended suggestions and group them by common themes after reviewing the answers.

```python
from pathlib import Path
import pandas as pd

# Read the cleaned survey data.
project_folder = Path.cwd().parent
clean_file = project_folder / 'data' / 'processed' / 'responses_clean.csv'
data = pd.read_csv(clean_file)

# Find the column that contains residents' suggestions.
suggestion_column = '15. Which local service needs the most urgent improvement, and what is your suggestion?'
suggestions = data[suggestion_column].dropna().astype(str).str.strip()
suggestions = suggestions[suggestions != '']

# Print the suggestions so they can be reviewed before grouping themes.
for number, suggestion in enumerate(suggestions, start=1):
    print(f'{number}. {suggestion}')
```
